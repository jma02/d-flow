```python
import polars as pl
import os
from IPython.display import display, Markdown
```

```python
run_name = "init"
profiling_dir = f"profiling/profiling-logs-{run_name}"
os.chdir(profiling_dir)
```

```python
df = pl.read_parquet("sample_images.parquet")

display(Markdown("## Top GPU Time Consumers"))
top_gpu = df.sort("Self_GPU_time_us", descending=True).head(10)
display(top_gpu)

# Filter specific operations
display(Markdown("## Convolution Operations"))
conv_ops = df.filter(pl.col("Name").str.contains("conv"))
display(conv_ops)

# Memory analysis
display(Markdown("## High Memory Usage Operations (>1MB)"))
high_memory = df.filter(pl.col("Self_GPU_memory_bytes") > 1_000_000)
display(high_memory)
```
