```python
import torch
import matplotlib.pyplot as plt
problem = 'circles'
data = torch.load(f'data/eit-{problem}-multiflow-128.pt')
test = data['test']
idx = 100
x = test['dtn_map'][idx].numpy()
y = test['media'][idx].numpy()
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].imshow(x)
axs[0].set_title('dtn_map')
axs[1].imshow(y)
axs[1].set_title('media')
plt.show()
```
