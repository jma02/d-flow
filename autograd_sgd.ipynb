```python
from typing import Callable, Dict, List, Optional

from torch import autocast
from torchdiffeq import odeint_adjoint 
import torch.nn as nn

def inverse_loss_fn(x, y):
    """
    Loss function for the inverse problem.
    """
    device = x.device
    if isinstance(x, np.ndarray):
        x= torch.tensor(x, device=device, dtype=torch.float32)
    if isinstance(y, np.ndarray):
        y = torch.tensor(y, device=device, dtype=torch.float32)

    x = x.float()
    y = y.float()


    loss = torch.mean((x - y) ** 2)

    return loss
    
def ode_integrate(ode_func: nn.Module,
                  init_x: torch.Tensor,
                  ode_opts: Dict = {},
                  t_eps: float = 0,
                  init_t: float = 0.,
                  final_t: float = 1.,
                  t_arr: Optional[List[float]] = None,
                  num_steps: int = 5,
                  intermediate_points: bool = False,
                  method: str = "euler"
      ) -> torch.Tensor:

    if t_arr is None:
        t = torch.linspace(init_t - t_eps, final_t, num_steps).to(init_x.device)
    else:
        t = torch.tensor(t_arr, dtype=torch.float32, device=init_x.device)

    ode_opts = {
        "atol": 1e-5,
        "rtol": 1e-5,
        "method": method,
        **ode_opts
    }

    z = odeint(
        func = lambda t, x: ode_func(x, t.expand(x.shape[0])),
        y0 = init_x,
        t = t,
        **ode_opts
    )

    if not intermediate_points:
        return z[-1]
    return z
```

```python
from torch.func import vmap, jvp, vjp
def sketch_jacobian(oversampling_param: int,
                    sketch_dim: int,
                    jac_dim: int,
                    model: nn.Module,
                    ode_integrate: Callable,
                    x0: torch.Tensor,
                    num_steps: int = 5,
                    intermediate_points: bool = False,
                    chunk_size: int = 8,
                    method: str = "euler"
                    ) -> torch.Tensor:
    # rangefinder step
    ell = sketch_dim + oversampling_param
    x0_detached = x0.detach().clone().requires_grad_(False)
    # we sample images of shape im_size x im_size and we compute jacobian vector products via jvp, then convert the outputs to vectors
    # those vectors are the columns of Y
    # we use jvp to form Y but we can easily also use finite difference
    def f(x_):
        x1 = ode_integrate(model, x_, num_steps=num_steps, intermediate_points=intermediate_points, method=method)
        # unnormalize
        x1 = 0.5 * (x1 + 1.0) * (train_max - train_min) + train_min
        return x1
        
    # we can do some vectorizing here using vmap
    Y = torch.zeros(jac_dim, ell).to(x0_detached.device)

    def single_jvp(omega):
        return jvp(f, (x0_detached,), (omega.unsqueeze(0),))[1].view(im_size*im_size)

    with torch.no_grad():
        omegas = torch.randn(ell, 1, im_size, im_size, device=x0.device)
        Y = vmap(single_jvp, chunk_size=8)(omegas).T  # shape: (jac_dim, ell)
        # compute QR 
        Q_range, _ = torch.linalg.qr(Y, mode='reduced')
        del Y, omegas

    with torch.no_grad():
        _, vjp_fn = vjp(f, x0_detached)
        Q_out = Q_range.reshape(im_size * im_size, ell)
        # transpose and reshape to images to pass into f
        Q_out_imgs = Q_out.T.reshape(ell, *f(x0_detached).shape)

        def apply_vjp(v_cotangent):
            # this should reshape back to a vector
            return vjp_fn(v_cotangent)[0].reshape(-1)

        # todo: probably need to make this dynamic
        B = vmap(apply_vjp, chunk_size=8)(Q_out_imgs)

        U_tilde, S, Vh = torch.linalg.svd(B, full_matrices=False)
        del B, Q_out_imgs
        U = Q_range @ U_tilde
        torch.cuda.empty_cache() if x0.device.type == 'cuda' else None
    return U, S, Vh
```

```python
num_steps = 5
intermediate_points = False
time_jac = time.time()
def f(x_):
    x1 = ode_integrate(model, x_, num_steps=num_steps, intermediate_points=intermediate_points, method=method)
    # unnormalize
    x1 = 0.5 * (x1 + 1.0) * (train_max - train_min) + train_min
    return x1
```

```python
# here is some code to just try running steepest descent
optim_steps = 10000
x0 = torch.randn((1, 1, im_size, im_size), device=device, dtype=torch.float32, requires_grad=True)
sketch_dim = 20
oversampling_param = 0
x1_traj = []
y = img_radon.clone().to(device)
eta = 0.25
losses = []
sketched_loss_grads = []
autograd_loss_grads = []
target_loss = .02
# Initialize progress bar
pbar = tqdm(range(optim_steps), desc="Steepest Descent")

for step in pbar:
    jac_dim = im_size * im_size
    U, S, Vh = sketch_jacobian(oversampling_param=oversampling_param, 
                                sketch_dim=sketch_dim, 
                                jac_dim=jac_dim, 
                                model=model, 
                                ode_integrate=ode_integrate, 
                                x0=x0, 
                                num_steps=5, 
                                intermediate_points=False,
                                chunk_size=None,
                                method=method
                                )
    # integrate x0
    x1 = ode_integrate(model, x0, num_steps=5, intermediate_points=False, method=method)
    x1 = 0.5 * (x1 + 1.0) * (train_max - train_min) + train_min
    # apply forward operator, << I'm not adding noise here
    x1_radon = radon.radon_transform(x1, N=5)
    loss = inverse_loss_fn(x1_radon, y)

    grad_x1 = torch.autograd.grad(inverse_loss_fn(x1_radon, y), x1, create_graph=True)[0]
    grad_x1 = grad_x1.view(im_size * im_size)
    # compute grad_{x_0} L(x_1) using chain rule
    # use element wise multiplication for the dot product <S, U^T @ grad_x1>
    loss_grad_sketch = (Vh.T @ (S * (U.T @ grad_x1)))

    loss_grad = torch.autograd.grad(loss, x0)[0]

    autograd_loss_grads.append(loss_grad)
    sketched_loss_grads.append(loss_grad_sketch)

    # update x0 using the computed gradient
    x0 = x0 - eta * loss_grad.view_as(x0)
    x1_traj.append(x1.detach().clone().cpu().numpy())
    # Update progress bar with loss
    pbar.set_postfix({'Loss': f'{loss.item():.6f}'})
    if step % 100 == 0:
        losses.append(loss)
    if loss < target_loss:
        print(f"Target loss reached at step {step}, Loss: {loss.item():.6f}")
        break
```

```python
# plot the norm of the loss grads over time
import matplotlib.pyplot as plt

autograd_norms = [torch.norm(g).item() for g in autograd_loss_grads]
print(sketched_loss_grads[0].shape)
print(autograd_loss_grads[0].shape)

# normalize the first gradients
fst_sketch = sketched_loss_grads[0] / torch.norm(sketched_loss_grads[0])
fst_autograd = autograd_loss_grads[0] / torch.norm(autograd_loss_grads[0])

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title('Normalized Sketched Gradient')
plt.imshow(fst_sketch.view(im_size, im_size).cpu().detach().numpy(), cmap='gray')
plt.colorbar()
plt.subplot(1, 2, 2)
plt.title('Normalized Autograd Gradient')
plt.imshow(fst_autograd.view(im_size, im_size).cpu().detach().numpy(), cmap='gray')
plt.colorbar()
plt.show()

# Plot gradient norms over time
sketched_norms = [torch.norm(g).item() for g in sketched_loss_grads]

plt.figure(figsize=(10, 6))
plt.plot(autograd_norms, label='Autograd', alpha=0.8)
plt.plot(sketched_norms, label='Sketched', alpha=0.8)
plt.yscale('log')
plt.xlabel('Iteration')
plt.ylabel('Gradient Norm')
plt.legend()
plt.title('Gradient Norms Over Time')
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.title('Difference (Autograd - Sketched)')
diff = (fst_autograd - fst_sketch.view(im_size, im_size)).cpu().detach().numpy()
plt.imshow(diff.squeeze(), cmap=cmocean.cm.diff, vmin=-np.abs(diff).max(), vmax=np.abs(diff).max())
plt.colorbar()

plt.subplot(1, 2, 2)
plt.title('Cosine Similarity Over Time')
cos_similarities = []
for i in range(len(sketched_loss_grads)):
    cos_sim = torch.cosine_similarity(
        sketched_loss_grads[i].flatten(), 
        autograd_loss_grads[i].flatten(), 
        dim=0
    ).item()
    cos_similarities.append(cos_sim)

plt.plot(cos_similarities)
plt.xlabel('Iteration')
plt.ylabel('Cosine Similarity')
plt.grid(True, alpha=0.3)
plt.show()
```
